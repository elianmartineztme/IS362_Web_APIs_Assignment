Introduction

The purpose of this assignment is to retrieve data from a New York Times API using Python and convert the JSON response into a pandas DataFrame. This notebook will demonstrate how to connect to the API, process the returned data, and organize the results into a table for easier analysis.

Tools Used

Python, Jupyter Notebook, pandas, and the New York Times API.

```python

import pandas as pd
import requests

print("Libraries imported successfully!")
```

```python

from getpass import getpass

# Request the API key without displaying it
api_key = getpass("Enter your NYT API key: ")

# Books API endpoint
url = "https://api.nytimes.com/svc/books/v3/lists/current/hardcover-fiction.json"

print("API connection settings are ready.")
```

```python

# Request data from the New York Times Books API
params = {"api-key": api_key}

try:
    response = requests.get(url, params=params, timeout=30)
    print("HTTP status code:", response.status_code)

    response.raise_for_status()

    data = response.json()
    print("API response status:", data.get("status"))
    print("Number of books:", len(data["results"]["books"]))

except requests.exceptions.RequestException as error:
    print("API request failed:", error)
```

```python

# Extract the book records from the JSON response
books = data["results"]["books"]

# Convert the JSON records into a pandas DataFrame
df = pd.DataFrame(books)

# Select useful columns for analysis
df = df[[
    "rank",
    "title",
    "author",
    "publisher",
    "weeks_on_list",
    "description"
]]

# Display the DataFrame
display(df)
```

## Methodology
I used the New York Times Books API to retrieve the current Hardcover Fiction bestseller list. First, I imported pandas and requests. I then used my API key to authenticate the request and retrieve the data in JSON format. After confirming that the request was successful with HTTP status code 200, I extracted the book records from the JSON response and converted them into a pandas DataFrame.

I selected six columns to make the results easier to understand: rank, title, author, publisher, weeks_on_list, and description.
## Results and Interpretation
The API returned 15 books from the Hardcover Fiction bestseller list. The DataFrame displays each book's ranking, title, author, publisher, number of weeks on the list, and a short description.

One interesting observation is that books can remain on the bestseller list for different lengths of time. For example, THE CALAMITY CLUB has been on the list for 27 weeks, while DEFY THE DUSK has been on the list for only one week. This shows how the DataFrame can be used to compare books based on their rankings and time on the bestseller list.
